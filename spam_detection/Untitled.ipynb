```python
nlp=spacy.load('en_core_web_sm')
def lemma(text):
    doc=nlp(''.join(text))
    tok=[token.lemma_ for token in doc if not token.is_punct and not token.is_stop]
    return ' '.join(tok)
```

```python
df['text']=df.text.apply(lemma)
```
